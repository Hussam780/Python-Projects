### 4. مفاهيم تقييم الأداء (Metrics Explanation)
يشرح المصدر منطق "مصفوفة الارتباك" (Confusion Matrix) ومقاييس التقرير النهائي [122، 123، 124]:

*   **True Positive (TP):** عندما يتنبأ الموديل بوجود المرض ويكون المريض مصاباً فعلاً.
*   **True Negative (TN):** عندما يتنبأ الموديل بعدم وجود المرض ويكون الشخص سليماً فعلاً.
*   **Precision (الدقة):** تقيس إجمالي عدد التنبؤات الصحيحة بالإصابة من بين جميع الحالات التي صنفها الموديل كمصابة.
*   **Recall (الاستدعاء):** يقيس مدى قدرة الموديل على اكتشاف جميع الحالات المصابة فعلياً من الداتا الأصلية.
*   **F1-Score:** يمثل المتوسط بين الـ Precision والـ Recall لتقديم تقييم متوازن.

```
from sklearn.datasets import load_breast_cancer
data_set = load_breast_cancer()
df = pd.DataFrame(data_set.data, columns=data_set.feature_names) # تحويلها لـ DataFrame
df['label'] = data_set.target # إضافة العمود المستهدف
```

*   **تقسيم ومعالجة البيانات:**

```
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

X = df.drop(columns='label', axis=1)
Y = df['label']
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=2)

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train) # توحيد مقاييس البيانات
X_test = scaler.transform(X_test)
```
